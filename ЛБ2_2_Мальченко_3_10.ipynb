Аналіз виживання залежно від статі: Обчисліть відсоток виживання для кожної статі. Чи була різниця у виживанні між чоловіками та жінками?

у жінок (1) шанс вижити був у рази вищим, ніж у чоловіків (0).

Обчисліть відсоток виживання для кожного класу (Pclass). Який клас мав найвищий рівень виживання (дати відповідь)?


```python
df.groupby('Pclass')['Survived'].mean() * 100
```

найвищий рівень виживання мав 1 клас

Визначте середній вік тих, хто вижив, і тих, хто не вижив. Чи впливає вік на виживання (дати відповідь)?


```python
df.groupby('Survived')['Age'].mean()
```

Середній вік пасажирів, які вижили, був трохи нижчим.

Розподіліть пасажирів на групи за рівнями тарифів (Fare) і обчисліть рівень виживання для кожної групи. Як тариф впливав на шанси виживання (дати відповідь)?

```python
df['FareGroup'] = pd.qcut(df['Fare'], 3, labels=['Низький', 'Середній', 'Високий'])
survival_by_fare = df.groupby('FareGroup', observed=False)['Survived'].mean() * 100
print(survival_by_fare)
```

 Чим вищим був тариф, тим більшими були шанси на виживання (від 19.8% у низькій групі до 55.9% у високій)

Аналіз класу та тарифу: Визначте середній тариф (Fare) для кожного класу (Pclass). Чи існує значна різниця у тарифах між класами (дати відповідь)?





```python
df.groupby('Pclass')['Fare'].mean()
```

Так, різниця існує: середній тариф 1-го класу у рази перевищував вартість квитків 2-го 3-го класів

Як пов'язано середній вік пасажирів з виживанням? Відповідь дати аналітично чи графічно


```python
mean_age = df.groupby('Survived')['Age'].mean()

plt.figure(figsize=(6, 4))
mean_age.plot(kind='bar', color=['#ff9999', '#66b3ff'])
plt.title('Середній вік за виживанням')
plt.xlabel('Чи вижив)')
plt.ylabel('Середній вік')
plt.xticks(ticks=[0, 1], labels=['Ні', 'Так)'], rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
