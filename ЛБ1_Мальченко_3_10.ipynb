До матриці:


додайте вектор:

```text
v = [10, 20, 30]
```

Проаналізуйте, як NumPy застосував broadcasting.

NumPy автоматично дублює вектор v вздовж рядків матриці A, додаючи значення до кожного відповідного стовпця всіх рядків завдяки сумісності їхніх останніх вимірів.

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]], dtype=float)

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)
````

````python
(A - mu) / sigma
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([[1,  2,  1],
              [2,  1, -1],
              [3, -1,  2]])
````

````python
b = np.array([9, 8, 3])
````

````python
solution_x = np.linalg.solve(A, b)
print(solution_x)
````

````python
calculated_b = A @ solution_x
print(calculated_b)
````

````python
is_equal = np.allclose(calculated_b, b)
print(is_equal)
````
